## Practice #1: 인터넷 쇼핑몰
- 인터넷 쇼핑몰에서 물건을 구입할 때, 구입액이 10만원 초과면 5%의 할인을 해준다 하자. 사용자에게 구입금액을 물어보고 최종적으로 할인금액과 지불금액을 출력하는 프로그램을 작성해보자.  
  (기본 if문 사용)
```
구입금액을 입력하세요: 100500
지불금액은
95475.0입니다.
```

## Practice #2: 수하물 비용 계산
- 항공사에서 짐을 부칠 때, 20kg이 넘어가면 20,000원을 내야 한다 하자. 20kg 이하면 수수료가 없다. 사용자로부터 짐의 무게를 입력받아 사용자가 지불해야 할 금액을 계산하는 프로그램을 작성해보자.  
  (if-else 사용)
```
짐의 무게는 얼마입니까? 18
짐에 대한 수수료가 없습니다.
감사합니다.

짐의 무게는 얼마입니까? 30
무거운 짐은 20000원을 내셔야 합니다.
감사합니다.
```

## Practice #3: 홀짝 구별
- 키보드에서 입력받은 정수가 홀수인지 짝수인지를 말해주는 프로그램을 작성해보자. 홀수와 짝수는 어떻게 구별할 수 있을까?  
  (if-else 사용)
```
정수를 입력하시오: 2
입력된 정수는 짝수입니다.

정수를 입력하시오: 5
입력된 정수는 홀수입니다.
```

## Practice #4: 큰 수 출력
- 사용자로부터 두 개의 정수를 입력받아서 둘 중에서 큰 수를 출력하는 프로그램을 작성해보자.  
  (if-else 사용)
```
첫번째 정수: 10
두번째 정수: 20
큰수는 20
```

## Practice #5: 졸업 학점 검사
- 어떤 대학교를 졸업하려면 적어도 140학점을 이수해야 하고 평점이 2.0은 되어야 한다 하자. 사용자에게 이수학점수와 평점을 입력받아 졸업 가능 여부를 출력하는 프로그램을 작성해보자.  
  (if-else 사용)
```
이수한 학점수를 입력하시오: 120
평점을 입력하시오: 2.3
졸업이 힘듭니다!
```

## Practice #6: 마트
- 마트에서 사과가 신선하면 사과를 사기로 한다. 만약 사과가 개당 1000원 미만이면 10개를 산다. 하지만 사과가 개당 1000원 이상이면 5개만 산다. 아래 출력결과를 참고하여 프로그램을 작성해보자.  
  (중첩 if-else 사용)
```
사과의 상태를 입력하세요: 신선
사과 1개의 가격을 입력하세요: 500
사과를 10개 산다.

사과의 상태를 입력하세요: 신선
사과 1개의 가격을 입력하세요: 1500
사과를 5개 산다.

사과의 상태를 입력하세요: 보통
사과를 사지 않는다.
```

## Practice #7: 음수, 0, 양수 구별
- 사용자로부터 정수를 받아서 음수, 0, 양수 중의 하나로 분류하는 프로그램을 작성해보자.

```
정수를 입력하시오: -10
입력된 정수는 음수입니다.

정수를 입력하시오: 0
입력된 정수는 0입니다.
```

## Practice #8: 윤년 판별
- 사용자로부터 연도를 입력받아서 해당 연도가 윤년인지 확인하여 결과가 윤년이면 “윤년”을, 아니면 “평년”을 출력하는 프로그램을 작성해보자.  
  (if-elif-else 사용)
- [힌트] 윤년 판별 공식
    - 연도가 4로 나누어 떨어지고 100으로 나누어 떨어지지 않으면 윤년
    - 또는 400으로 나누어 떨어지면 윤년
```
연도를 입력하세요: 2011
평년입니다.

연도를 입력하세요: 2012
윤년입니다.
```
