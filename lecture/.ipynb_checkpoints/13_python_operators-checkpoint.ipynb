## 연산자

### 산술연산자(Arithmetic operator)
- 수식(expression): 피연산자들과 연산자의 조합
- 연산자(operator): 연산을 나타내는 기호
- 피연산자(operand): 연산의 대상이 되는 값

- 산술연산자의 종류
    - 피연산자 모두 정수형이 아닐 경우, 결과는 항상 실수
| 산술연산자 | 설명 | 사용예 | 예설명 |
|---|---|---|---|
| = | 대입연산자 | a = 3 | 정수 3을 a에 대입 |
| + | 더하기 | a = 5 + 3 | 5와 3을 더한 값을 a에 대입 |
| - | 빼기 | a = 5 - 3 | 5에서 3을 뺀 값을 a에 대입 |
| * | 곱하기 | a = 5 * 3 | 5와 3을 곱한 값을 a에 대입 |
| / | 나누기 | a = 5 / 3 | 5를 3으로 나눈 값을 a에 대입 |
| // | 나누기(몫) | a = 5 // 3 | 5를 3으로 나눈 뒤 소수점을 버린 값을 a에 대입 |
| % | 나머지값 | a = 5 % 3 | 5를 3으로 나눈 뒤 나머지값을 a에 대입 |
| ** | 제곱 | a = 5 ** 3 | 5의 3제곱을 a에 대입 |

- 나눗셈의 몫 연산
    - `//`: 나눗셈의 몫
    - `/`: 나눗셈(결과는 항상 실수)

```python
7 / 4  # 실수 나눗셈
```

```python
7 // 4  # 정수 나눗셈
```

```python
7.0 / 4.0  # 피연산자가 실수인 경우
```

```python
7.0 // 4.0
```

- 나머지연산자
    - `%` 연산자
    - 나눗셈이 수행된 후 나머지값을 결과로 산출
        - 예) 숫자가 짝수(홀수)인지 나머지연산자를 사용하여 확인한다.
    - 짝수 % 2는 항상 0
    - 홀수 % 2는 항상 1

```python
e = 7 % 4
print(e)
```

```python
n1 = 30
n2 = 35
print(n1 % 2)  # 짝수
print(n2 % 2)  # 홀수
```

### 복합연산자
- 연산과 할당을 동시에 표현
- `+=`처럼 대입연산자와 다른 연산자를 합쳐 놓은 연산자

```python
val1 = 3
val1 = val1 + 2
print(val1)
```

```python
val1 = 3
val1 += 2
print(val1)
```

### 관계연산자(Relational operator)
- 관계연산자란?
    - 두 값 중 어떤 것이 큰지, 작은지, 같은지를 비교하는 것
    - 결과는 참(True)이나 거짓(False)
    - 주로 조건문(if)이나 반복문(for, while)에서 사용

- 관계연산자의 종류
| operator | description | example |
|---|---|---|
| == | equal | 5 == 7  # False |
| != | not equal | 5 != 7  # True |
| > | greater than | 5 > 7  # False |
| < | less than | 5 < 7  # True |
| >= | greater than or equal | 5 >= 7  # False |
| <= | less than or equal | 5 <= 7  # True |

- 관계연산자의 결과값
    - 참(True)이나 거짓(False)를 생성

100 1

- 관계연산자 예
    - a와 b를 비교하기 위한 관계연산자 `==`
    - `=`을 하나만 쓰는 경우 → 오류 발생

```python
a = 100; b = 100
print(a == b)  # 두 수가 같으면 True
print(a != b)
print(a > b, a < b)
print(a <= b, a >= b)
```

- 주의
    - a = 10: 수 a에 10을 할당 (오른쪽에서 왼쪽으로!)
    - a == 10: 수 a의 값이 10과 같은지 비교

### 논리연산자(Logical operator)
- 논리연산자의 종류
| 논리연산자 | 의미 | 설명 | 사용예 |
|---|---|---|---|
| and | ~이고, 그리고 (AND) | 둘 다 참이어야 참 | (a > 100) and (a < 200) |
| or | ~이거나, 또는 (OR) | 둘 중 하나만 참이어도 참 | (a == 100) or (a == 200) |
| not | ~아니다, 부정(NOT) | 참이면 거짓, 거짓이면 참 | not(a < 100) |

- 진리표
| 입력값 A | 입력값 B | A 그리고 B (A and B) | A 또는 B (A or B) | A가 아니다 (!A) |
|---|---|---|---|---|
| True | True | True | True | False |
| True | False | True | False | False |
| False | True | False | True | True |
| False | False | False | False | True |

- 논리연산자 예

```python
a = 320
print(True and a == 300)  # True and False
print((a > 100) and (a % 2 == 0))  # True and True
print((a == 100) or (a == 200))  # False or False
print(not(a == 100))
```

- 복합조건 예
    - 나이가 10살 이상(조건 1)이고(and) 키가 165 이상이면(조건 2) 놀이기구를 탈 수 있다.
        - (age >= 10) and (height >= 165)
    - 나이가 10살 이상(조건 1)이거나(or) 키가 165 이상이면(조건 2) 놀이기구를 탈 수 있다.
        - (age >= 10) or (height >= 165)

### 연산자우선순위
- 같은 우선순위를 갖는 연산자는 왼쪽부터 계산
- 단, `**` 연산자는 오른쪽부터 계산(예: 2 ** 2 ** 3 = 2 ** 8 = 256)
| 우선순위 | 연산자 | 설명 |
|---|---|---|
| Highest | ( ) | anything in brackets is done first |
|  | ** | exponentiation |
|  | -x, +x | arithmetic operators |
|  | *, /, %, // | arithmetic operators |
|  | +, - | arithmetic operators |
|  | <, >, <=, >=, !=, == | relational operators |
|  | =, +=, -=, *=, etc | assignment operators |
|  | not | logical operator |
|  | and | logical operator |
| Lowest | or | logical operator |
