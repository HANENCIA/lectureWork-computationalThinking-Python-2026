## Practice #1: 과일가게
- 연세 과일가게에서 사과를 1개에 1,000원, 배 1개에 2,000원, 멜론은 1개에 3,000원에 판매하고 있다. 그런데 오늘은 사과가 시들어서 원래 가격의 10%를 할인해서 팔려고 한다. 고객들이 구매한 과일 총액을 계산하여 정수로 출력하는 프로그램을 작성해보자.
```
사과 갯수를 입력하세요: 10
배 갯수를 입력하세요: 10
멜론 갯수를 입력하세요: 10
전체금액은: 59000
```

````python
apple = int(input('사과 갯수를 입력하세요: '))
pear = int(input('배 갯수를 입력하세요: '))
melon = int(input('멜론 갯수를 입력하세요: '))

apple_total = apple * 1000 * 0.9  # 실수형으로 변환됨
pear_total = pear * 2000
melon_total = melon * 3000
total = apple_total + pear_total + melon_total
total = int(total)  # 실수값을 정수형으로 변환

print('전체금액은: ', total)
````

## Practice #2: BMI 계산하기
- 사용자로부터 신장과 체중을 입력받아서 BMI 값을 출력하는 프로그램을 작성해보자.
```
몸무게를 kg 단위로 입력하시오: 85.0
키를 미터 단위로 입력하시오: 1.83
당신의 BMI = 25.381468541909282
```

````python
weight = float(input("몸무게를 kg 단위로 입력하시오: "))
height = float(input("키를 미터 단위로 입력하시오: "))
bmi = (weight / (height**2))
print("당신의 BMI =", bmi)
````

## Practice #3: 분초계산하기
- 초 단위의 시간을 입력받아서 몇 분 몇 초인지를 계산하는 프로그램을 작성해보자.
```
초 단위 시간을 입력하세요: 1000
16분 40초
```

````python
sec = int(input("초 단위 시간을 입력하세요: "))
min = sec // 60
remainder = sec % 60
print(min, "분", remainder, "초")
````
