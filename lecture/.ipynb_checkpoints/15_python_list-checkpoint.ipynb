### 리스트 인덱싱
- 인덱스 값을 통해 리스트가 가리키고 있는 값에 접근
- 인덱싱 범위
    - Positive indexing: 0 ~ n-1
    - Negative indexing: -1 ~ -n
    - n: 리스트의 크기(len() 함수를 통해 구함)
- 대괄호 [] 안에 인덱스 번호를 입력하여 특정 요소값을 참조

````python
letters = ['P', 'y', 't', 'h', 'o', 'n']
n = len(letters)  # size of list, n=6
print(n)
````

- 범위를 벗어나면 IndexError 발생
- 정수가 아닌 인덱스를 사용하면 TypeError 발생

````python
letters[6]  # IndexError (out of range)
````

````python
letters[4.0]  # TypeError
````

### 복잡한 리스트 인덱싱

````python
a = [2020, ['월요일', '컴퓨팅사고'], '2-3교시']

print(a[0])  # 2020
print(a[1])  # ['월요일', '컴퓨팅사고']
print(a[1][0])  # '월요일'
print(a[1][1])  # '컴퓨팅사고'
print(a[2])  # '2-3교시'
````

### 인덱싱을 이용한 리스트 원소 수정

````python
# 원소를 수정하기
gold = [0, 4, 5, 0, 1, 1, 5, 4, 4, 1]
print(gold)
````

### 리스트 슬라이싱
- 리스트의 일부 원소들을 추출하는 방법
- 슬라이싱의 결과는 리스트(원소가 없거나 1개일 경우도 리스트)
- 형식
    - L[begin:end:step]  # n개의 요소를 갖는 리스트 L

- 슬라이싱 규칙(begin, end ≥ 0이며 begin < end인 경우로 설명)
    - step > 0
        - 인덱스를 begin부터 end-1까지 step씩 증가하며 리스트 참조
        - begin을 생략하면 0부터, end를 생략하면 끝까지, step을 생략하면 +1씩
```
L[:end:step] L[begin::step] L[begin:end]
```
    - step < 0
        - 인덱스를 begin부터 end+1까지 |step|씩 감소하며 리스트 참조
        - begin을 생략하면 끝에서부터, end를 생략하면 첫 번째 원소까지
```
L[:end:step] L[begin::step]
```

````python
A = [0, 1, 2, 3, 4, 5, 6, 7, 8]
````

````python
print(A[2:5:1]) # [2, 3, 4] A[2:5]와 동일
print(A[:5]) # [0, 1, 2, 3, 4] A[0:5:1]과 동일
print(A[4:])  # [4, 5, 6, 7, 8]
print(A[:6:2])  # [0, 2, 4] A[0], A[2], A[4]를 선택
print(A[1:7:2])  # [1, 3, 5] A[1], A[3], A[5]를 선택
print(A[4:4])  # [] 빈 리스트
print(A[4:5])  # [4]
print(A[4:-1])  # [4, 5, 6, 7] A[4:8]과 동일

# 비교
print(A[4])  # 4 원소 참조, 슬라이싱 아님
````

### 슬라이싱을 이용한 원소 교체/추가/제거
- 리스트에서 하나의 원소 수정하기 → 인덱싱 사용
- 연속된 범위값 수정하기 → 슬라이싱 사용
    - 주의: 교체/추가 시 오른쪽 항은 반드시 리스트여야 함  
        - (A[2:3] = 5 → 오류)

````python
A = [0, 1, 2, 3, 4]
````

````python
A[1:3] = [1]  # [0, 1, 3, 4] A[1]과 A[2]를 1로 교체
print(A)
A[2:3] = [5, 6]  # [0, 1, 5, 6, 4] A[2]를 5, 6으로 교체
print(A)
A[1:1] = ['a', 'b']  # [0, 'a', 'b', 1, 5, 6, 4] 인덱스 1에 'a', 'b' 추가(끼워넣기)
print(A)
A[len(A):len(A)] = [3]  # [0, 'a', 'b', 1, 5, 6, 4, 3] A 리스트 끝에 새 원소 A[7]를 추가
print(A)
A[4:5] = []  # [0, 'a', 'b', 1, 6, 4, 3] A[4] 제거
print(A)
A[1:3] = []  # [0, 1, 6, 4, 3] A[1], A[2] 제거
print(A)
del A[:2]  # [6, 4, 3] A[0], A[1] 제거
print(A)


# 비교
A[1] = []  # [6, []] A[1]에 empty list []를 할당
print(A)
````

### 리스트의 연산: +, *
- Concatenation operator +: 리스트의 연결 연산
    - 리스트들을 연결하여 새로운 리스트로 만듦
- Repetition Operator *: 리스트의 반복 연산
    - list * n: list의 item을 n번 반복하여 새로운 list를 만듦

````python
L1 = [1, 3, 5]
L2 = [2, 4, 6]
````

````python
L3 = L1 + L2  # [1, 3, 5, 2, 4, 6] 리스트 L1과 L2를 연결한 리스트 반환
print(L3)
````

````python
L4 = L1 * 3  # [1, 3, 5, 1, 3, 5, 1, 3, 5] 리스트 L1을 3회 반복한 리스트 반환
print(L4)
````

````python
print(L1, L2)  # [1, 3, 5][2, 4, 6] 리스트 L1과 L2는 변하지 않음
````

### 리스트의 연산: in, not in
- Membership operator in, not in
    - item in list
        - item이 list에 있으면 True, 없으면 False
    - item not in list
        - item이 list에 없으면 True, 있으면 False

````python
A = [1, 3, 5, 7, 9]
````

````python
print(3 in A)  # True
print(6 in A)  # False
print(2 not in A)  # True
print(5 not in A)  # False
````

### 리스트의 메소드
- 메소드(method)
    - 데이터 객체에 대해 어떤 일을 처리할 수 있도록 하는 코드
- 리스트에서 사용할 수 있는 메소드
    - 리스트.메소드의 형태로 사용
    - 리스트 메소드 목록

````python
print(dir(list))
````

### 리스트의 메소드 목록
| method | Description (x: 리스트, a: 임의의 object) |
|---|---|
| x.append(a) | 데이터 a를 리스트 x의 끝에 추가 |
| x.extend(L) | 리스트 L의 모든 원소를 리스트 x의 마지막에 추가 |
| x.insert(i, a) | a를 리스트 x의 인덱스 i에 추가 |
| x.remove(a) | 리스트 x에서 원소값이 데이터 a인 첫 원소 제거(반환값 없음) |
| x.pop() | 리스트 x의 마지막 원소 제거 및 반환 |
| x.pop(i) | 리스트 x의 인덱스 i의 요소를 제거하고 그 값을 반환 |
| x.clear() | 리스트 x의 모든 원소를 삭제(빈 리스트가 됨) |
| x.index(a) | 리스트 x에서 원소값이 a인 첫 번째 원소의 인덱스를 반환 |
| x.count(a) | 리스트 x에서 a 값과 같은 원소의 개수를 반환 |
| x.sort() | x의 원소들을 오름차순으로 정렬 내림차순으로 정렬하려면 x.sort(reverse=True) 사용 |
| x.reverse() | x의 원소들을 역으로 재배치(정렬과 다름) |
| x.copy() | a shallow copy of the list (y = x[:]와 동일) |

### 리스트 관련 내장함수
| function | Description (인수 x: 리스트) |
|---|---|
| len(x) | x의 원소 개수 반환 |
| list(y) | Iterable한 y를 리스트로 변환하여 반환 |
| max(x) | x의 원소 중 최댓값 반환(비교 불가이면 TypeError) |
| min(x) | x의 원소 중 최솟값 반환(비교 불가이면 TypeError) |
| sorted(x) | x의 원소를 정렬한 리스트 반환 x는 불변이고 오름차순 또는 내림차순으로 정렬 (내림차순: sorted(x, reverse=True)) |
| sum(x) | x의 원소 합을 반환(더할 수 없으면 Type Error) |

````python
list_a = [1, 2, 3, 4, 5]
list_b = [6, 7, 8, 9, 10]
````

````python
list_a.append(6)
print(list_a)
````

````python
list_a.pop()
print(list_a)
````

````python
list_a.insert(-1,5.5)
print(list_a)
````

````python
list_a.pop(4)
print(list_a)
````

````python
list_a.extend(list_b)
print(list_a)
````

````python
list_a.reverse()
print(list_a)
````

````python
print(sum(list_a))
````

````python
list_c = list(range(1, 10, 2))
print(list_c)
````
