## 선 그래프 (line graph)

````python
plt.plot([10, 20, 30, 40])
plt.show()
````

````python
plt.plot([1, 2, 3, 4], [12, 43, 25, 15])
plt.show()
````

### 축(axis) 관련

#### 축 레이블

````python
x = [1, 2, 3, 4, 5, 6]
y = [12, 43, 25, 15, 27, 32]
plt.plot(x, y)
plt.xlabel('X-Label', labelpad=10)  # x-axis의 Label, labelpad=축 레이블의 여백(padding)
plt.ylabel('Y-Label', labelpad=10)  # Y-axis의 label, labelpad=축 레이블의 여백(padding)
plt.show()
````

````python
x = [1, 2, 3, 4, 5, 6]
y = [12, 43, 25, 15, 27, 32]
plt.plot(x, y)
plt.xlabel('X-Label', labelpad=10, loc='right')  # loc=x-axis Label의 위치: 'left’, ‘center’, ‘right’
plt.ylabel('Y-Label', labelpad=10, loc='top')  # loc=y-axis Label의 위치: ‘bottom’, ‘center’, ‘top’
plt.show()
````

#### 축의 범위

````python
x = [1, 2, 3, 4, 5, 6]
y = [12, 43, 25, 15, 27, 32]

plt.plot(x, y)

plt.xlim([0, 5])  # X축의 범위: [xmin, xmax]
plt.ylim([0, 20])  # Y축의 범위: [ymin, ymax]

plt.show()
````

````python
x = [1, 2, 3, 4, 5, 6]
y = [12, 43, 25, 15, 27, 32]

plt.plot(x, y)

plt.axis([0, 5, 0, 20])  # X, Y축의 범위: [xmin, xmax, ymin, ymax]

plt.show()
````

### 제목, 범례(legend) 추가
- 범례
| 원하는 위치 | 코드 |
|---|---|
| 자동으로 가장 적합한 곳 | `loc='best'` |
| 오른쪽 위 | `loc='upper right'` |
| 왼쪽 위 | `loc='upper left'` |
| 왼쪽 아래 | `loc='lower left'` |
| 오른쪽 아래 | `loc='lower right'` |
| 가운데 | `loc='center'` |
| 위 가운데 | `loc='upper center'` |
| 아래 가운데 | `loc='lower center'` |
| 왼쪽 가운데 | `loc='center left'` |
| 오른쪽 가운데 | `loc='center right'` |

````python
plt.title('Title')  # Title

plt.plot([10, 20, 30, 40], label='asc')
plt.plot([40, 30, 20, 10], label='desc')

plt.legend()  # 범례 (legend)
plt.show()
````

### line 색상 변경
![image](./plot_color.png)
- 색상은 css 색상을 사용할 수 있음

````python
plt.title('Color')

plt.plot([10, 20, 30, 40], color='skyblue', label='skyblue')
plt.plot([40, 30, 20, 10], color='pink', label='pink')

plt.legend()
plt.show()
````

### 선 모양(Linestyle) 변경
![image](./plot_linestyle_1.png)
![image](./plot_linestyle_2.png)
- 선 종류는 위와 같이 쓸 수 있음

````python
plt.title('Linestyle')

plt.plot([10, 20, 30, 40], color='r', linestyle='--', label='dashed')
plt.plot([40, 30, 20, 10], color='g', ls=':', label='dotted')

plt.legend()
plt.show()
````

### 마커 모양(Marker) 변경
![image](./plot_marker.png)
- 마커 모양은 위와 같이 쓸 수 있음

````python
plt.title('Marker')

plt.plot([10, 20, 30, 40], 'r.', linestyle="--", label='circle')
plt.plot([40, 30, 20, 10], 'g^', label='triangle up')

plt.legend()
plt.show()
````

### Grid 설정

````python
plt.plot([10, 20, 30, 40], 'r.', linestyle="--", label='circle')
plt.plot([40, 30, 20, 10], 'g^', label='triangle up')
plt.grid(True)

plt.show()
````

````python
plt.plot([10, 20, 30, 40], 'r.', linestyle="--", label='circle')
plt.plot([40, 30, 20, 10], 'g^', label='triangle up')
plt.grid(True, axis='x')

plt.show()
````

````python
plt.plot([10, 20, 30, 40], 'r.', linestyle="--", label='circle')
plt.plot([40, 30, 20, 10], 'g^', label='triangle up')
plt.grid(True, axis='y')

plt.show()
````

````python
plt.plot([10, 20, 30, 40], 'r.', linestyle="--", label='circle')
plt.plot([40, 30, 20, 10], 'g^', label='triangle up')
plt.grid(True, axis='y', color='blue', alpha=0.5, linestyle='--')

plt.show()
````

## 히스토그램 (histogram)

````python
plt.hist([1, 1, 2, 3, 4, 5, 6, 6, 7, 8, 10])
plt.show()
````

### 주사위 시뮬레이션
    1. 주사위를 굴린다
    2. 나온 결과를 기록한다
    3. 1-2번 과정을 n번 반복한다
    4. 주사위의 눈이 나온 횟수를 히스토그램으로 그린다

- 1-6 사이의 랜덤숫자 만들기
```
import random
print(random.randint(1, 6))
```

````python
import random

dice = []
for i in range(5):
    dice.append(random.randint(1, 6))

plt.hist(dice, bins=[0.5, 1.5, 2.5, 3.5, 4.5, 5.5, 6.5])
# bin(빈): 히스토그램에서 데이터를 묶는 구간
# bins=6: bin을 6개 만들것. 즉, Matplotlib이 데이터 범위를 6등분함
# 다만 주사위 데이터에서는 경계가 예쁘게 1, 2, 3... 중앙에 오지 않을 수 있음
 
plt.xticks(range(1, 7))  # x축: 1, 2, 3, 4, 5, 6
plt.yticks(range(0, 6))  # y축: 0, 1, 2, 3, 4, 5
plt.show()
````

## 산점도 (scatterplot)

````python
import matplotlib.pyplot as plt
import numpy as np

np.random.seed(0)

n = 50
x = np.random.rand(n)
y = np.random.rand(n)

plt.scatter(x, y)
plt.show()
````

````python
import matplotlib.pyplot as plt
import numpy as np

np.random.seed(0)

n = 50
x = np.random.rand(n)
y = np.random.rand(n)
area = (30 * np.random.rand(n))**2
colors = np.random.rand(n)

plt.scatter(x, y, s=area, c=colors) # s: Size, c: Color
plt.show()
````

## 파이 차트 (Pie Chart)

````python
ratio = [34, 32, 16, 18]
labels = ['Apple', 'Banana', 'Melon', 'Grapes']

plt.pie(ratio, labels=labels, autopct='%.1f%%')
plt.show()
````

````python
ratio = [34, 32, 16, 18]
labels = ['Apple', 'Banana', 'Melon', 'Grapes']
explode = [0, 0.10, 0, 0.10] # explode에 적용: Banana와 Grapes에 한해 반지름보다 약 10% 벗어나도록 그림

plt.pie(ratio, labels=labels, autopct='%.1f%%', startangle=260, explode=explode)
plt.show()
````

## Box Plot

````python
df = pd.read_csv("extremum_20261116133126.csv", encoding="euc-kr")
df
````

````python
plt.boxplot(df["평균기온(℃)"].tolist(), label="평균기온")
plt.rc('font', family='Malgun Gothic')  # 한국어 Font를 지정해주지 않으면 깨짐
plt.legend()
plt.show()
````

````python
plt.boxplot([df["최고기온(℃)"].tolist(), df["평균기온(℃)"].tolist()], tick_labels=["최고기온(℃)", "평균기온(℃)"])
plt.rc('font', family='Malgun Gothic')  # 한국어 Font를 지정해주지 않으면 깨짐
plt.rcParams['axes.unicode_minus'] = False  # 맑은 고딕에 -가 없기 때문에 대체
plt.show()
````

## 응용

````python
x = range(df.shape[0])
labels = df["일시"].tolist()

plt.title('Color')

plt.plot(df["최고기온(℃)"].tolist(), color='skyblue', label='최고기온')
plt.plot(df["평균기온(℃)"].tolist(), color='pink', label='최저기온')
plt.xticks(ticks=x, labels=labels, rotation=90)
plt.legend(loc="upper right")
plt.savefig('graph_result.png')
plt.show()
````
