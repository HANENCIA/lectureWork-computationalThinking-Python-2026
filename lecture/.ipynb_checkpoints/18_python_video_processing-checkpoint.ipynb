# 영상 처리
- 영상은 이미지가 시간 순서대로 이어진 것이므로 보통 다음 과정으로 진행
    1. 영상 파일 열기
    2. 프레임(이미지)을 하나씩 읽기
    3. 각 프레임에 이미지 처리 적용
    4. 처리 결과를 화면에 보여주거나 새 영상 파일로 저장하기

## 라이브러리 (OpenCV) 설치
- OpenCV는 영상과 이미지를 처리하기 위한 대표적인 라이브러리
- 주요 기능
    - MP4, AVI 등 다양한 영상 파일 읽기 및 저장
    - 프레임 단위 이미지 처리
    - 크기 변경, 회전, 자르기
    - 흐림, 엣지 검출, 그레이스케일 변환
    - 텍스트, 도형, 사각형 그리기

## 라이브러리 불러오기

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

## 영상 파일 확인

```python
video_path = 'video.mp4'

cap = cv2.VideoCapture(video_path)

if not cap.isOpened():
    print('영상 파일을 열 수 없습니다. video.mp4 파일이 있는지 확인하세요.')
else:
    print('영상 파일을 정상적으로 열었습니다.')
```

## 영상 정보 확인

- `cap.get(cv2.CAP_PROP_FRAME_WIDTH)`: 영상 가로 크기
- `cap.get(cv2.CAP_PROP_FRAME_HEIGHT)`: 영상 세로 크기
- `cap.get(cv2.CAP_PROP_FPS)`: 초당 프레임 수(FPS)
- `cap.get(cv2.CAP_PROP_FRAME_COUNT)`: 전체 프레임 수

```python
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = cap.get(cv2.CAP_PROP_FPS)
frame_count = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

print(f'가로: {width}')
print(f'세로: {height}')
print(f'FPS: {fps:.2f}')
print(f'전체 프레임 수: {frame_count}')
print(f'영상 길이: {frame_count / fps:.2f}초')
```

## 첫 번째 프레임 읽기
- `cap.read()`는 한 번 호출할 때마다 다음 프레임을 읽음
- 반환값은 다음 두 개임
    - `ret`: 프레임을 정상적으로 읽었는지 나타내는 값
    - `frame`: 읽어 온 이미지 프레임
- OpenCV는 색상을 **BGR** 순서로 저장함. 따라서 Matplotlib으로 보기 위해서는 BGR을 RGB로 변환하는 작업이 필요

```python
ret, frame = cap.read()

if ret:
    frame_rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
    plt.figure(figsize=(8, 6))
    plt.imshow(frame_rgb)
    plt.axis('off')
    plt.title('First Frame')
    plt.show()
else:
    print('프레임을 읽을 수 없습니다.')
```

## 프레임 자르기
- frame[세로 시작:세로 끝, 가로 시작:가로 끝]

```python
cropped = frame[height // 4 : height * 3 // 4, width // 4 : width * 3 // 4]

cropped_rgb = cv2.cvtColor(cropped, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(6, 6))
plt.imshow(cropped_rgb)
plt.axis('off')
plt.title('Cropped Frame')
plt.show()
```

## 프레임 크기 변경
- cv2.resize(이미지, (가로, 세로))

```python
resized = cv2.resize(frame, (320, 240))

resized_rgb = cv2.cvtColor(resized, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(5, 4))
plt.imshow(resized_rgb)
plt.axis('off')
plt.title('Resized Frame')
plt.show()
```

## 그레이스케일 변환
- cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

```python
gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

plt.figure(figsize=(8, 6))
plt.imshow(gray, cmap='gray')
plt.axis('off')
plt.title('Grayscale Frame')
plt.show()
```

## 흐림(Blur) 효과
- cv2.GaussianBlur(이미지, (커널 가로, 커널 세로), 표준편차)
    - 커널 크기가 커질수록 영상이 더 흐려짐
    - 커널 크기는 보통 홀수 사용

```python
blur = cv2.GaussianBlur(frame, (15, 15), 0)

blur_rgb = cv2.cvtColor(blur, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(8, 6))
plt.imshow(blur_rgb)
plt.axis('off')
plt.title('Gaussian Blur')
plt.show()
```

## 가장자리 검출(Edge Detection)
- Canny 알고리즘은 이미지에서 물체의 경계, 즉 가장자리를 찾음
- cv2.Canny(이미지, 낮은 임계값, 높은 임계값)
- 일반적으로 그레이스케일 이미지에 적용함

```python
edges = cv2.Canny(gray, 100, 200)

plt.figure(figsize=(8, 6))
plt.imshow(edges, cmap='gray')
plt.axis('off')
plt.title('Canny Edge Detection')
plt.show()
```

## 프레임에 텍스트 넣기
- cv2.putText(이미지, 문구, 위치, 글꼴, 크기, 색상, 두께)
- OpenCV에서 색상 순서는 BGR
- 예를 들어 `(0, 255, 0)`은 초록색

```python
text_frame = frame.copy()

cv2.putText(text_frame,
            'OpenCV Video Processing',
            (50, 80),
            cv2.FONT_HERSHEY_SIMPLEX,
            1.5,
            (0, 255, 0),
            3)

text_rgb = cv2.cvtColor(text_frame, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(8, 6))
plt.imshow(text_rgb)
plt.axis('off')
plt.title('Text on Frame')
plt.show()
```

## 프레임에 사각형 그리기

- cv2.rectangle(이미지, 시작 좌표, 끝 좌표, 색상, 두께)
- 두께가 -1이면 사각형 내부를 채움

```python
rect_frame = frame.copy()

cv2.rectangle(rect_frame,
              (width // 4, height // 4),
              (width * 3 // 4, height * 3 // 4),
              (255, 0, 0),
              5)

rect_rgb = cv2.cvtColor(rect_frame, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(8, 6))
plt.imshow(rect_rgb)
plt.axis('off')
plt.title('Rectangle on Frame')
plt.show()
```

## 여러 효과 비교하기

```python
plt.figure(figsize=(14, 10))

plt.subplot(2, 2, 1)
plt.imshow(cv2.cvtColor(frame, cv2.COLOR_BGR2RGB))
plt.title('Original')
plt.axis('off')

plt.subplot(2, 2, 2)
plt.imshow(gray, cmap='gray')
plt.title('Grayscale')
plt.axis('off')

plt.subplot(2, 2, 3)
plt.imshow(cv2.cvtColor(blur, cv2.COLOR_BGR2RGB))
plt.title('Blur')
plt.axis('off')

plt.subplot(2, 2, 4)
plt.imshow(edges, cmap='gray')
plt.title('Edges')
plt.axis('off')

plt.tight_layout()
plt.show()
```

## 처리한 영상 저장하기
처리한 프레임들을 새로운 MP4 파일로 저장하기
- cv2.VideoWriter(파일명, 코덱, FPS, (가로, 세로))
- `VideoWriter`는 BGR 색상 프레임을 기대하므로, 그레이스케일 이미지를 다시 BGR로 변환해야 함

```python
output_path = 'gray_video.mp4'

out = cv2.VideoWriter(output_path,
                      cv2.VideoWriter_fourcc(*'mp4v'),
                      fps,
                      (width, height))

cap.set(cv2.CAP_PROP_POS_FRAMES, 0)

for i in range(frame_count):
    ret, frame = cap.read()

    if not ret:
        break

    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    gray_bgr = cv2.cvtColor(gray, cv2.COLOR_GRAY2BGR)

    out.write(gray_bgr)

out.release()

print(f'{output_path} 파일로 저장했습니다.')
```

## 영상 자원 해제
영상 처리를 끝낸 뒤에는 반드시 `release()`를 호출하여 파일 핸들을 닫아야 함

```python
cap.release()
print('영상 파일을 닫았습니다.')
```
