## 변수(Variable)

### 변수 소개
- 변수는 값을 저장하는 메모리 공간
- 값을 저장하는 상자 또는 그릇으로 생각할 수 있음

### 변수 생성
- Python에서 변수는 값이 할당되는 순간에 생성
- 대입연산자와 변수의 위치

````python
score = 20
score
````

````python
print(score)
````

- 대입연산자와 변수의 위치가 틀린 경우
    - 대입연산자 왼쪽에 넣을 그릇이 없는 경우

````python
10 = 100
````

### 변수사용: 값의 저장공간
- 생성된 변수에는 얼마든지 다른 값을 저장할 수 있음

````python
score = 20
score = 30
score
````

### 변수사용: 한변수를 다른변수에 복사-저장

- 다른변수의 값을 변수에 넣기
- 변수 var2의 값(200)을 복사하여 변수 var1에 대입

````python
var2 = 200
var1 = var2
print(var1)
````

### 변수사용: 두값의 연산결과 저장
- 숫자끼리의 계산결과를 변수에 넣기

````python
var1 = 100 + 100
print(var1)
````

- 내부적으로 먼저 계산
    1. 100 + 100 = 200
    2. 계산결과를 대입

### 변수사용
- 변수와 숫자의 계산결과를 변수에 넣기

````python
var2 = 200
var1 = var2 + 100
print(var1)
````

- 내부적으로 먼저 계산
    1. var2==200 + 100 = 300
    2. 계산결과를 대입

### 변수사용: 한변수를 여러개의 변수로 연속저장
- 연속된 값을 대입

````python
var1 = var2 = var3 = var4 = 100
print(var1, var2, var3, var4)
````

````python
var4 = 100
var3 = var4
var2 = var3
var1 = var2
print(var1, var2, var3, var4)
````

### 변수사용: 변수와 값의 연산결과를 같은 변수에 저장
- 변수 자신의 값에 계산결과를 대입

````python
var1 = 100
var1 = var1 + 200
print(var1)
````

````python
var1 = 100
var1 += 200
print(var1)
````

### 변수를 활용하는 목적: 반복되는 값의 저장공간
- 다음과 같이 Turtle 그래픽을 사용하여 반지름이 100인 3개의 원을 그리는 프로그램이 있다고 하자

````python
import turtle

# Session Clear
try:
    turtle.bye()
except Exception:
    pass

turtle.TurtleScreen._RUNNING = True

t = turtle.Turtle()
t.shape("turtle")
radius = 100
t.circle(radius)
# 반지름이 100인 원이 그려진다.
t.fd(30)
t.circle(radius)
# 반지름이 100인 원이 그려진다.
t.fd(30)
t.circle(radius)  # 반지름이 100인 원이 그려진다
turtle.done()
````

### 변수의 유용성: 반복되는 값을 변경하고자 할 경우
- 하지만 갑자기 원의 반지름을 50으로 변경하여 다시 그려야 한다면?  
  - 원의 반지름이 변수로 표현되었기 때문에 변수값만 변경

````python
import turtle

# Session Clear
try:
    turtle.bye()
except Exception:
    pass

turtle.TurtleScreen._RUNNING = True

t = turtle.Turtle()
t.shape("turtle")
radius = 50
t.circle(radius)
# 반지름이 50 원이 그려진다.
t.fd(30)
t.circle(radius)
# 반지름이 50인 원이 그려진다.
t.fd(30)
t.circle(radius)  # 반지름이 50인 원이 그려진다
turtle.done()
````

## 자료형

### 변수의 자료형
- 자료의 종류(자료형)란?
- 변수에 어떤 종류의 자료라도 담을 수 있음
- 변수의 자료형은 담는 자료의 종류에 따라 결정됨
- 변수의 자료형은 값을 넣는 순간마다 변경될 수 있음
- `type()` 함수는 변수의 자료형을 확인하는 함수

| 변수 | 설명 |
|---|---|
| bool | 불리언형 변수 |
| int | 정수형 변수 |
| float | 실수형 변수 |
| str | 문자열 변수 |

### 불리언자료형
- 참(True)이나 거짓(False)만 저장
- 불리언형은 단독으로 사용하기보다는 if 조건문이나 while 반복문 등과 함께 주로 사용

````python
a = True
print(type(a))
````

````python
a = (100 == 100)
b = (10 > 100)
print(a, b)
````

### 숫자자료형
- 정수형(int)
  - 소수점이 없는 데이터
  - 예: 100, -123, 0 등
- 실수형, 부동소수점(float)
  - 실수형은 3.14, -2.7처럼 소수점이 있는 데이터

````python
a = 123
print(type(a))
````

````python
a = 3.14
b = 3.14e5
print(type(b))
print(a, b)
````

### 숫자자료형
- 산술연산 수행 가능
  - 사칙연산 +, -, *, / 를 수행
  - 지수계산 **, 나머지를 구하는 %, 나눈 후에 소수점을 버리는 // 연산

````python
a = 10
b = 20
print(a + b, a - b, a * b, a / b)
````

````python
a, b = 9, 2
print(a ** b, a % b, a // b)
````

### 문자열자료형
- 문자열은 양쪽을 큰따옴표(" ")나 작은따옴표(' ')로 감싸야 함

````python
a = "파이썬만세"
a
````

````python
print(type(a))
````

````python
print("작은따옴표는' 모양입니다.")
````

````python
print('큰따옴표는" 모양입니다.')
````

### 자료형변환
- 원하는 자료형으로 변환 가능
- int() 함수  
  - `var = int(var)` ⇒ var에 해당하는 변수를 정수형으로 변환
- float() 함수  
  - `var = float(var)` ⇒ var에 해당하는 변수를 실수형으로 변환
- str() 함수  
  - `var = str(var)` ⇒ var에 해당하는 변수를 문자열로 변환

### 자료형변환: int() 함수
- 실수값을 정수로 변환: 소수점 아래는 버림
- 정수형태의 문자열을 정수로 변환 가능
- 실수형태의 문자열은 정수로 변환 불가능

````python
a = int(5.6)
# 소수점 뒤를 버리고 정수로 변환
c = int('15')
# 정수형태의 문자열을 정수로 변환
print(a, c)
````

````python
score = '85'  # 정수형태의 문자열
score + 5  # TypeError 발생!!
````

````python
int(score) + 5
````

### 자료형변환: float() 함수
- 정수값을 실수로 변환
- 실수형태의 문자열을 실수로 변환

````python
x = float(3)
y = float(15.7)
z = x + z
print(x, y, z)
````

````python
print(type(x), type(y), type(z))
````

````python
x = 3
y = 15.7
z = x+z
print(x, y, z)
print(type(x), type(y), type(z))
````

- 참고: 부동소수점은 약간의 오차를 가진다

````python
a = 0.1 + 0.2
b = 0.3

print(a)          # 0.30000000000000004
print(a == b)     # False
````

### 자료형변환: str() 함수
- 정수값을 문자열로 변환
- 실수값을 문자열로 변환

````python
age = 21
print(age, '살')
# 값과 값 사이에 공백이 출력됨
````

````python
s = str(age)
print(s + '살')
# 값과 값 사이에 공백 없이 출력
````

````python
print(str(age) + '살')   # 값과 값 사이에 공백 없이 출력
````
