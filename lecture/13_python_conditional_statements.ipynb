## 조건문

### 조건문
- 문제를 해결할 때 어떤 조건에 따라서 두 개 이상의 실행 경로 가운데 하나를 선택해야 하는 경우가 발생
- 눈이 오는가?

### 기본 if문
- if 조건문 형식
- 조건식이 참(True)이면 실행할 문장이 처리되고
- 거짓(False)이면 아무것도 실행하지 않고 프로그램을 종료
```
if condition:
    statement_true
else:
    statement_true
```

### 기본 if문
- if 조건문 형식
```python
if condition:
    statements
statementA
```
- statementA는 if 문과 관계없는 statement
    - 들여쓰기를 안 했기 때문
    - if 문 실행 후 바로 실행됨

### 기본 if문 예제
- 정수를 하나 입력받아 100보다 큰 수인지 작은 수인지 판별하여 100보다 작으면 “100보다 작다”를 출력하라.

````python
num = int(input("정수를 입력하시오: "))
if num < 100:
    print("100보다 작다")
````

### if-else 문
- 조건이 거짓(False)일 때도 수행할 문장이 있을 경우 사용
```python
if condition:
    statements
else:
    statements
```

### if-else 문 예제
- 정수를 하나 입력받아 100보다 큰 수인지 작은 수인지 판별하여 100보다 작으면 “100보다 작다”를, 그렇지 않으면 “100보다 크거나 같다”를 출력하는 코드를 작성하라.

````python
num = int(input("정수를 입력하시오: "))
if num < 100:
    print("100보다 작다")
else:
    print("100보다 크거나 같다")
````

```
score >= 90
if:
    print("합격입니다.")
    print("장학금도 받을 수 있습니다.")
```
- 블록은 특정한 동작을 위해서 코드가 모여 있는 상태
- 블록은 들여쓰기로 구분
- Python은 들여쓰기 자체가 문법
- 들여쓰기는 코드를 읽기 쉽도록 일정한 간격을 띄워서 작성하는 방법
    - 공백(스페이스), 탭(tab) 등 여러 가지 방법이 있으나 탭 사용을 권장

### 중첩 if-else 문
- if-else 문장 안에 또 다른 if-else 문장을 넣을 수 있음
```
if condition1:
    if condition2:
        statement1
    else:
        statement2
else:
    if condition3:
        statement3
    else:
        statement4
```

````python
appleQuality = "신선"
applePrice = 500

if appleQuality == "신선":
    if applePrice < 1000:
        print("10개를 산다")
    else:
        print("5개를 산다")
else:
    print("사과를 사지 않는다.")
````

### if-elif-else 문
- 조건에 따라 다중으로 분기되는 경우
- elif는 이전 조건문이 거짓일 때 수행됨
- else는 불필요하면 빼도 됨

```
if condition1:
    statement1
elif condition2:
    statement2
elif condition3:
    statement3
else:
    statement4
```

### if-elif-else 문 예제
- 학생들의 성적을 받아서 성적이 90점 이상이면 A학점, 80점 이상이고 90점 미만이면 B학점, 70점 이상이고 80점 미만이면 C학점을 출력

````python
score = int(input("점수를 입력하세요: "))
if score >= 90:
    print("A")
elif score >= 80:  # 80 <= score < 90
    print("B")
elif score >= 70:  # 70 <= score < 80
    print("C")
elif score >= 60:  # 60 <= score < 70
    print("D")
else:
    # score < 60
    print("F")
print("학점입니다.")
````

### if-elif-else 문
- 다음 두 코드는 동일한 일을 수행

````python
score = 82
if score >= 90:
    print("A")
elif score >= 80:
    print("B")
elif score >= 70:
    print("C")
elif score >= 60:
    print("D")
else:
    print("F")
print("학점입니다.")
````

````python
score = 82
if score >= 90:
    print("A")
else:
    if 80 <= score:
        print("B")
    else:
        if 70 <= score:
            print("C")
        else:
            if 60 <= score:
                print("D")
            else:
                print("F")
print("학점입니다.")
````
