## Practice #2: 합계 구하기

| i의 값 | sum의 값 |

| 1번째 반복 | 1 | 0+1 |
| 2번째 반복 | 2 | 0+1+2 |
| 3번째 반복 | 3 | 0+1+2+3 |
| … | … | … |
| 10번째 반복 | 10 | 0+1+2+3+…+10 |
