## Practice #1: 제곱값을 반환 함수
- 정수를 입력받아서 제곱한 값을 반환하는 square() 함수를 작성해보자.
- 입력값, 반환값 모두 존재
```
정수를 입력하시오: 10
제곱값은 100
```

````python
def square(n):
    return n * n

num = int(input("정수를 입력하시오: "))
print("제곱값은", square(num))
````

## Practice #2: 큰 수 반환 함수
- 두 개의 정수를 입력받아 더 큰 수를 찾아서 반환하는 get_max() 함수를 작성해보자.
- 입력값, 반환값 모두 존재
```
첫번째 정수를 입력하시오: 10
두번째 정수를 입력하시오: 20
큰수는 20
```

````python
def get_max(x, y):
    if x > y:
        return x
    else:
        return y

a = int(input("첫번째 정수를 입력하시오: "))
b = int(input("두번째 정수를 입력하시오: "))
print("큰수는", get_max(a, b))
````

## Practice #3: 거리 반환 함수
- 인자(x, y)를 입력하면 (0, 0)으로부터의 거리를 구하는 distance() 함수를 작성해보자.
- 입력값은 없고 반환값만 존재
```
x 좌표 입력: 3
y 좌표 입력: 4
거리: 5.0
```

````python
def distance():
    x = int(input("x 좌표 입력: "))
    y = int(input("y 좌표 입력: "))
    c = (x**2 + y**2) ** 0.5
    return c

print("거리: ", distance())
````

## Practice #4: 배수 찾기 함수
- 사용자로부터 배수와 최댓값을 입력받고, 최댓값까지의 모든 배수를 출력하는 mul(n, max_num) 함수를 작성하고 테스트해보자.
- 단, 최댓값으로 배수와 같거나 배수보다 작은 숫자가 전달되면 "잘못된 입력입니다."를 출력하도록 하시오. (입력값만 존재)
```
배수 입력: 6
최댓값 입력: 35
6
12
18
24
30

배수 입력: 7
최댓값 입력: 5
잘못된 입력입니다.
```

````python
def mul(n, max_num):
    if n >= max_num:
        print("잘못된 입력입니다.")
    else:
        for i in range((int)(max_num / n)):
            print(n * (i + 1))

n = int(input('배수 입력: '))
max_num = int(input('최댓값 입력: '))
mul(n, max_num)
````

## Practice #5: 햄버거가게
- 간단한 햄버거가게 메뉴 선택 시스템을 구현하여보자.
- 메뉴를 화면에 출력하는 함수 print_menu(): 입력값, 반환값 모두 없음
- 사용자로부터 입력받은 메뉴 번호가 메뉴 선택 범위를 벗어나지 않았는지 검사하는 함수 check_menu(): 입력값만 존재
```
1. 치즈버거세트
2. 불고기버거세트
3. 치킨버거세트
4. 종료
메뉴 선택: 2
2번 메뉴가 선택되었습니다.

1. 치즈버거세트
2. 불고기버거세트
3. 치킨버거세트
4. 종료
메뉴 선택: 5
잘못된 메뉴를 선택하셨습니다.
```

````python
def print_menu():
    print("1. 치즈버거세트")
    print("2. 불고기버거세트")
    print("3. 치킨버거세트")
    print("4. 종료")


def check_menu(menu):
    if menu < 1 or menu > 4:
        return 1
    else:
        return 0


menu = 0
while menu != 4:
    print_menu()
    menu = int(input('메뉴 선택: '))
    if check_menu(menu):
        print("잘못된 메뉴를 선택하셨습니다.")
    elif menu != 4:
        print(menu, "번 메뉴가 선택되었습니다.")
````
