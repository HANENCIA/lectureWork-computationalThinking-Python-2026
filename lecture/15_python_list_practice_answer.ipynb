## Practice #1: 리스트 연습
- 비어 있는 리스트를 만들고, 키보드로부터 5개의 정수 데이터를 입력받은 다음에 리스트에 추가하고, 리스트 내용을 출력한 후, 평균을 계산하여 출력해보자.
```
Enter 5 integers: 10
Enter 5 integers: 20
Enter 5 integers: 30
Enter 5 integers: 40
Enter 5 integers: 50
[10, 20, 30, 40, 50]
평균: 30.0
```

````python
datalist = []
total = 0
avg = 0

for i in range(5):
    indata = int(input("Enter 5 integers: "))
    datalist.append(indata)
    total = total + indata

print(datalist)
avg = total / 5
print("평균:", avg)
````

## Practice #4: 리스트 메소드 연습
- 학생이 3명 [아이유, 폴킴, 황인욱]인 과에 청하가 편입을 했다.
    1. "청하"를 리스트에 추가하라.
    2. 동명이인 폴킴이 새로 편입하여 "폴킴" 뒤에 "폴킴"을 추가하라.
    3. "폴킴"이 몇 명인지 구해보아라.
    4. "황인욱"이 다른 과로 전과를 하였다. "황인욱"을 찾아 삭제하라.
    5. 현재 리스트에서 두 번째 학생이 자퇴를 하였다. 두 번째 학생을 삭제하라.
    6. 현재 출석부를 내림차순으로 정렬하라.
```
(1) => ['아이유', '폴킴', '황인욱', '청하']
(2) => ['아이유', '폴킴', '폴킴', '황인욱', '청하']
(3) => 2
(4) => ['아이유', '폴킴', '폴킴', '청하']
(5) => ['아이유', '폴킴', '청하']
(6) => ['폴킴', '청하', '아이유']
```

````python
students = ["아이유", "폴킴", "황인욱"]
students.append("청하")
print("(1) =>", students)

idx = students.index("폴킴")
students.insert(idx, "폴킴")
print("(2) =>", students)

cnt = students.count("폴킴")
print("(3) =>", cnt)

students.remove("황인욱")
print("(4) =>", students)

del students[1]
print("(5) =>", students)

students.reverse()
print("(6) =>", students)
````

### Practice #5: 자동차 회사
- 한 회사의 차 라인업을 모두 입력하려고 한다. 회사에서 개발한 차종은 나온 순서대로 다음과 같다.
    1. 이 회사의 차종에 "Merce"를 추가하라.
    2. "Lex" 차종이 단종되었다. 자료에서 이것을 삭제하라.
    3. "Gene" 2019년형이 새로 추가되었다. 자료에서 이것을 추가하라.
    4. 두 번째부터 네 번째 개발한 모델까지 찾아서 출력하라.
```
현재 차종: ['Gene', 'Lex', 'Infini', 'Lambor', 'Linc']
(1) Merce 차종 추가: ['Gene', 'Lex', 'Infini', 'Lambor', 'Linc', 'Merce']

### Practice #6: 방번호
- 세아는 지안이의 옆집에 새로 이사왔다. 세아는 자기 방번호를 예쁜 플라스틱 숫자로 문에 붙이려고 한다.
- 세아의 옆집에서는 플라스틱 숫자를 한 세트로 판다. 한 세트에는 0번부터 9번까지 숫자가 하나씩 들어 있다.
- 세아의 방번호가 주어졌을 때, 필요한 세트의 개수의 최소값을 출력하는 프로그램을 작성해보자.
- 단, 6은 9를 뒤집어서 이용할 수 있고, 9는 6을 뒤집어서 이용할 수 있다. max() 함수는 리스트의 최대값을 반환한다.
```
방번호 입력: 9999
2
```

````python
room = input('방번호 입력: ')
arr = [0] * 9

for i in room:
    if i == '9':
        arr[6] += 1
    else:
        arr[int(i)] += 1

arr[6] = (arr[6] + 1) // 2
print(max(arr))
````
