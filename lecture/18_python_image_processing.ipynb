## 이미지 필터링
- Pillow 패키지는 이미지를 필터링하기 위한 기본적인 필터들이 있음
- 이미지 필터 사용법
    - Filter(필터종류)를 호출하여 사용
    - ImageFilter 모듈 사용
- 필터종류
    - BLUR: 흐림
    - Contour: 윤곽
    - Detail: 세부묘사
    - EDGE_ENHANCE: 가장자리강조
    - EMBOSS: 엠보싱효과
    - FIND_EDGES: 모서리찾기
    - SHARPEN: 선명도효과
    - SMOOTH: 부드러운

## Blur 효과

````python
from PIL import Image, ImageFilter
````

````python
im = Image.open('mail.png')
blurImage = im.filter(ImageFilter.BLUR)
blurImage.save('blur.png')
blurImage.show()
````

### 윤곽선(Contour) 효과

````python
im = Image.open('mail.png')
contourImage = im.filter(ImageFilter.CONTOUR)
contourImage.save('contour.png')
contourImage.show()
````

### 엠보싱(Emboss) 효과

````python
im = Image.open('mail.png')
embossImage = im.filter(ImageFilter.EMBOSS)
embossImage.save('emboss.png')
embossImage.show()
````

### 가장자리 검출(Find Edges)

````python
im = Image.open('mail.png')
edgesImage = im.filter(ImageFilter.FIND_EDGES)
edgesImage.save('edges.png')
edgesImage.show()
````

## 그레이스케일(Grayscale) 이미지 변환
픽셀별 RGB 값을 가중합하여 회색조로 변환
```
(red * 0.1111) + (green * 0.004) + (blue * 0.885)
```

````python
def convert_grayscale(image):
    width, height = image.size
    new_image = Image.new('RGB', (width, height), 'white')

    for i in range(width):
        for j in range(height):
            pixel = image.getpixel((i, j))
            red = pixel[0]
            green = pixel[1]
            blue = pixel[2]
            gray = red * 0.111 + green * 0.004 + blue * 0.885
            new_image.putpixel((i, j), (int(gray), int(gray), int(gray)))

    return new_image

original = Image.open('mail.png')
output = convert_grayscale(original)
output.show()
````

## 밝기 증가
- 각 RGB 채널에 100을 더해 이미지를 밝게 만듦

````python
from PIL import Image

def brighten(image):
    width, height = image.size
    new_image = Image.new('RGB', (width, height), 'white')

    for i in range(width):
        for j in range(height):
            pixel = image.getpixel((i, j))
            red = min(pixel[0] + 100, 255)
            green = min(pixel[1] + 100, 255)
            blue = min(pixel[2] + 100, 255)
            new_image.putpixel((i, j), (red, green, blue))

    return new_image

original = Image.open('mail.png')
output = brighten(original)
output.show()
````
